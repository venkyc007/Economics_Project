# College Effects on Academic Outcomes: Regression Analysis

## Overview

This analysis estimates **college fixed effects** on student outcomes using OLS regression. The core question: *How much does attending a specific college affect your grade 12 English performance, after controlling for your prior ability?*

## The Model

### Basic Specification

```
englishmarks_12 = β₀(Bin) + β₁(marks) + Σ δⱼ(College_j) + ε
```

Where:
- **englishmarks_12**: Student's grade 12 English score (outcome)
- **marks**: Entrance exam score (ability control)
- **College_j**: Indicator = 1 if student attended college j, 0 otherwise
- **δⱼ**: College fixed effect (the causal parameter of interest)

### What This Estimates

Each **δⱼ** represents: *"The expected change in English marks from attending college j instead of the baseline college, holding entrance marks constant."*

For example:
- If δⱼ = +5, students at college j score 5 marks higher than similar students at the baseline
- If δⱼ = -3, students score 3 marks lower

## Data

- **Sample**: 6,348 students in residential colleges
- **Outcome**: Grade 12 English marks (mean = 77.5, range = 5-98)
- **Treatment**: 193 college×stream combinations
- **Control**: Entrance exam marks (mean = 37.1, range = -5 to 108.75)

### Key Variables
```
englishmarks_12    - Grade 12 English score
marks              - Entrance exam score (10th grade)
allotted_JC_code   - College×stream assigned (e.g., "501502_bpc")
```

## Identification Strategy

### The Challenge: Selection Bias

Students aren't randomly assigned to colleges. Higher-ability students attend better colleges, creating **omitted variable bias**:

```
Better students → Better colleges → Better outcomes        
```

Simply comparing average outcomes across colleges would **overestimate** good college effects.

### The Solution: Controlling for Prior Ability

By dividing students into 'pair-bins', we compare students with **similar prior ability** across colleges:

'Pair-bins' are determined through the following criterion:
- If a student has listed both college A and B in preferences, and hass A>B as preference (A ranked higher than B)
- And if a student falls into a particular 'marks' bin of width **5**

Then he is alloted into a unique pair-bin with ID- AxB_MarksRange
```
OLS Assumption: Conditional on these pair-bins, college assignment
                is "as-if random" with respect to unobserved ability
```

This is a **selection-on-observables** strategy. It works if:
1. Entrance marks capture most ability differences
2. Other existing confounders (family background, motivation, etc.) are homogenised within the pair-bin
## Model Specification

### Baseline Model (Model 1)
```
Y = Bin FEs + College FEs + ε
(or)
Y(demeaned)= College FEs + ε
```

**Why entrance marks?**
- Strongest predictor of future performance (r = 0.41 with English marks)
- Main criterion for college assignment
- Captures student ability at time of allocation

**Why college fixed effects?**
- Captures all college-specific factors (teaching quality, peer effects, resources)
- Non-parametric: doesn't assume linear college quality
- Interpretation: direct comparison to baseline college

### Model Fit
```
R² = 0.2189
```

**Interpretation**:
- Colleges + entrance marks explain ~22% of variation in English scores
- Most variation remains unexplained (student effort, teacher effects, etc.)

### Ability Coefficient

β₁ = 0.308 (SE = 0.017, p < 0.001)

**Interpretation**:
- 1 mark higher on entrance → 0.31 marks higher in grade 12 English
- This is the **within-college** relationship
- Highly significant, confirming entrance marks capture ability


### College Fixed Effects

**Distribution**:
- Range: [-24.7, +10.9] marks
- Mean absolute effect: |δⱼ| = 5.2 marks
- Standard deviation: 6.8 marks

**Significant Effects**:
- 135 colleges significantly different from baseline (p < 0.05)
- 69 negative effects, 66 positive effects

## Limitations

1. **Omitted Variables**: Family background, motivation, health not controlled
2. **Selection Bias**: Students may choose colleges for unobserved reasons
3. **Single Outcome**: Only English marks, not overall performance
4. **No Causal Mechanism**: Can't distinguish teaching quality vs peer effects vs resources
## Conclusion

This regression provides **descriptive evidence** that colleges differ substantially in their association with student outcomes, even after controlling for prior ability. The effects range from -25 to +11 marks on a 100-point scale, suggesting college choice matters.

However, without random assignment or stronger identification (e.g., discontinuities, instruments), we cannot rule out selection bias. These estimates are best interpreted as **upper bounds** on true causal effects.

For causal inference, see the matching analysis (M3), which uses a more credible identification strategy.
