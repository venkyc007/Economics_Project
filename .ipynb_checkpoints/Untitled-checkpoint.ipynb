```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Set seed for reproducibility
np.random.seed(42)

# 1. Mapping File
mapping = pd.DataFrame({
    'suspect_id': ['S1', 'S2', 'S3', 'S4'],
    'name': ['Alex', 'Priya', 'Sameer', 'Nisha'],
    'device_mac': ['MAC_001', 'MAC_002', 'MAC_003', 'MAC_004'],
    'card_id': ['CARD_101', 'CARD_102', 'CARD_103', 'CARD_104']
})
mapping.to_csv('mapping.csv', index=False)

# 2. WiFi Heartbeats (~400 rows)
# Pings every 2 minutes for 9 hours
wifi_times = pd.date_range(start='2025-10-14 18:00:00', end='2025-10-15 03:00:00', freq='2min')
wifi_data = []

for t in wifi_times:
    for mac in ['MAC_001', 'MAC_003', 'MAC_004']:
        # Random background noise for others
        loc = np.random.choice(['Hallway_AP', 'Library_AP', 'Cafeteria_AP', 'Gym_AP'])
        wifi_data.append([mac, t.isoformat(), loc, np.random.randint(-80, -40)])
    
    # Priya's logic: At Lab until midnight, then phone turns off (no rows), then reappears at exit
    if t < pd.Timestamp('2025-10-15 00:01:00'):
        loc = 'Lab_AP' if t > pd.Timestamp('2025-10-14 23:45:00') else 'Hallway_AP'
        wifi_data.append(['MAC_002', t.isoformat(), loc, -45])
    elif t > pd.Timestamp('2025-10-15 02:05:00'):
        wifi_data.append(['MAC_002', t.isoformat(), 'Exit_Gate_AP', -50])

pd.DataFrame(wifi_data, columns=['mac_address', 'timestamp', 'ap_name', 'signal_strength_dbm']).to_csv('wifi_heartbeats.csv', index=False)

# 3. Access Logs (~150 rows)
# Swiping in/out of various rooms to create noise
access_data = []
rooms = ['Cafeteria', 'Library_Internal', 'Gym', 'Supply_Closet', 'Server_Room']

for cid in ['CARD_101', 'CARD_103', 'CARD_104']:
    # Each person does ~40 random swipes throughout the night
    temp_time = pd.Timestamp('2025-10-14 18:00:00')
    while temp_time < pd.Timestamp('2025-10-15 02:00:00'):
        room = np.random.choice(rooms)
        access_data.append([cid, temp_time.isoformat(), room, 'enter'])
        temp_time += pd.Timedelta(minutes=np.random.randint(10, 30))
        access_data.append([cid, temp_time.isoformat(), room, 'exit'])
        temp_time += pd.Timedelta(minutes=np.random.randint(5, 15))

# Priya's critical path (The Culprit)
access_data.append(['CARD_102', '2025-10-14T18:30:00', 'Building_Main_Entry', 'enter'])
access_data.append(['CARD_102', '2025-10-14T19:00:00', 'Cafeteria', 'enter'])
access_data.append(['CARD_102', '2025-10-14T19:45:00', 'Cafeteria', 'exit'])
access_data.append(['CARD_102', '2025-10-14T23:58:00', 'Lab_Room_302', 'enter']) # The Crime
access_data.append(['CARD_102', '2025-10-15T02:15:00', 'Building_Main_Entry', 'exit'])

# Sameer's Red Herring (Entering the side building but not the lab)
access_data.append(['CARD_103', '2025-10-14T23:45:00', 'Building_Side_Entry', 'enter'])
access_data.append(['CARD_103', '2025-10-15T00:15:00', 'Building_Side_Entry', 'exit'])

access_df = pd.DataFrame(access_data, columns=['card_id', 'timestamp', 'location', 'event'])
access_df.sort_values('timestamp').to_csv('access_logs.csv', index=False)

# 4. GPS Data (~250 rows)
gps_data = []
gps_times = pd.date_range(start='2025-10-14 18:00:00', end='2025-10-15 03:00:00', freq='5min')

for t in gps_times:
    # Alex (MAC_001) - Fake Alibi at Home. 
    # Coordinates 19.1450, 72.9500 are ~5km away, but accuracy is so bad it overlaps with the lab
    gps_data.append(['MAC_001', t.isoformat(), 19.1450, 72.9500, np.random.randint(400, 1200)])
    
    # Nisha (MAC_004) - At the Library
    gps_data.append(['MAC_004', t.isoformat(), 19.1332, 72.9185, np.random.randint(10, 25)])
    
    # Priya (MAC_002) - Precise until she disappears
    if t < pd.Timestamp('2025-10-15 00:01:00'):
        gps_data.append(['MAC_002', t.isoformat(), 19.1330, 72.9180, np.random.randint(5, 12)])
    elif t > pd.Timestamp('2025-10-15 02:10:00'):
        gps_data.append(['MAC_002', t.isoformat(), 19.1320, 72.9170, 10])

pd.DataFrame(gps_data, columns=['device_mac', 'timestamp', 'lat', 'lon', 'accuracy_m']).to_csv('gps_data.csv', index=False)

print("Professional Dataset Generated: mapping.csv, wifi_heartbeats.csv, access_logs.csv, gps_data.csv")
```
