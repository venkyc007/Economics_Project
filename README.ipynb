# College Value-Added Analysis
### Overview
This project estimates the causal effect of attending different colleges on student academic outcomes using different econometric models. The core question: Does the college you attend matter for your academic performance, after accounting for your prior ability? Furthermore, do colleges that are considered better (ranked higher in the preference list) provide greater academic value to students?
### Methodology
#### The Problem
Students don't randomly attend colleges- they're assigned based on entrance exam scores and preferences. This creates selection bias: better students attend better colleges, making it hard to isolate the college's true effect.
#### The Solution
We will account for the fixed effects of these heterogenous factors in different models, with varying degrees of complexity, and weights and biases. They will all act as robustness checks to ensure the conclusions we obtain are independent of the unobservable heterogeneities present between students. 
Three independent sets of regression have been run with different underlying methodologies to generate sets of 'value-add coefficients' of each college. These have then been compared amongst each other to test the validity of data found.
#### The Result
The comparisons on average show a correlation factor of approximately 0.6 to 0.7 between different sets of coefficients, indicating moderately strong correlation and confirming that college itself does play a factor in the academic outcome of the student.
